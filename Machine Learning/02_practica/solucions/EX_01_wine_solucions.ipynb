### Exercici 1 — Explora Wine

Carrega el dataset amb `load_wine(as_frame=True)` i munta un DataFrame `dades` amb
les 13 columnes químiques més una columna `classe` amb el nom de la varietat
(`wine.target_names[wine.target]`; descarta la columna `target` del `.frame`, que fa
la mateixa feina en números).

Digues quantes files i columnes té, quantes classes hi ha i si estan equilibrades.
Fes un `describe()` i fixa't en el rang (mínim i màxim) de `proline` i de
`nonflavanoid_phenols`.

*Com saps que ho has fet bé: 178 files, 14 columnes (13 + `classe`), 3 classes
(59 / 71 / 48 mostres). `proline` va de 278 a 1680; `nonflavanoid_phenols` va de
0,13 a 0,66 — mateix dataset, escales completament diferents.*

```python
# Exercici 1
from sklearn.datasets import load_wine

wine = load_wine(as_frame=True)
dades = wine.frame.drop(columns="target").copy()
dades["classe"] = wine.target_names[wine.target]

print("Files i columnes:", dades.shape)
print("\nMostres per classe:")
print(dades["classe"].value_counts())

descripcio = dades.describe()
print("\nRang de proline:", descripcio.loc["min", "proline"], "-",
      descripcio.loc["max", "proline"])
print("Rang de nonflavanoid_phenols:", descripcio.loc["min", "nonflavanoid_phenols"],
      "-", descripcio.loc["max", "nonflavanoid_phenols"])
descripcio
```

### Exercici 2 — El flux complet

Separa `dades` en `X` (les 13 columnes químiques) i `y` (la columna `classe`). Fes
el split 70/30 amb `train_test_split(..., test_size=0.3, random_state=42,
stratify=y)`, entrena un `DecisionTreeClassifier(random_state=42)` i mesura la
precisió a l'examen.

*Com saps que ho has fet bé: al voltant del 96 % de precisió (96,3 % amb aquesta
llavor).*

```python
# Exercici 2
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

cols = wine.feature_names

X = dades[cols].to_numpy(dtype=float)
y = dades["classe"].to_numpy()

X_entrena, X_examen, y_entrena, y_examen = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

arbre = DecisionTreeClassifier(random_state=42)
arbre.fit(X_entrena, y_entrena)
precisio_arbre = arbre.score(X_examen, y_examen)
print(f"Precisió de l'arbre: {precisio_arbre:.1%}")
```

### Exercici 3 — k-NN sense escalar i escalat

Amb el mateix `X_entrena`/`X_examen` de l'exercici 2 (sense tocar-los), entrena un
`KNeighborsClassifier(n_neighbors=5)` **sense escalar** i anota la precisió a
l'examen. Després aplica `StandardScaler` (ajustat només amb `X_entrena`) i entrena
un altre `KNeighborsClassifier(n_neighbors=5)` amb les dades escalades.

Escriu en un comentari, amb les teves paraules, per què hi ha tanta diferència —
relaciona-ho amb el parany de les escales que vau veure amb Iris al quadern de k-NN.

*Com saps que ho has fet bé: sense escalar la precisió ronda el 72 %; escalant puja
fins a prop del 94 %. Més de 20 punts de diferència, i no has canviat cap idea del
model.*

```python
# Exercici 3
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

knn_sense = KNeighborsClassifier(n_neighbors=5)
knn_sense.fit(X_entrena, y_entrena)
precisio_knn_sense = knn_sense.score(X_examen, y_examen)

escalador = StandardScaler()
X_entrena_esc = escalador.fit_transform(X_entrena)
X_examen_esc = escalador.transform(X_examen)

knn_esc = KNeighborsClassifier(n_neighbors=5)
knn_esc.fit(X_entrena_esc, y_entrena)
precisio_knn_esc = knn_esc.score(X_examen_esc, y_examen)

print(f"Sense escalar: {precisio_knn_sense:.1%}")
print(f"Escalat:       {precisio_knn_esc:.1%}")

# k-NN només calcula distàncies. "proline" arriba a valer 1680 i
# "nonflavanoid_phenols" no passa de 0,66: la resta de columnes pràcticament
# no compten en la distància fins que totes queden a la mateixa escala amb
# StandardScaler.
```

### Exercici 5 — Compara els cinc models

Sobre Wine **escalat** (`X_entrena_esc`/`X_examen_esc` de l'exercici 3), entrena els
cinc models que coneixeu: `KNeighborsClassifier(n_neighbors=5)`,
`DecisionTreeClassifier(random_state=42)`,
`RandomForestClassifier(n_estimators=100, random_state=42)`,
`LogisticRegression(random_state=42, max_iter=10000)` i `SVC(random_state=42)`.
Munta una taula (`DataFrame`) amb el model i la seva precisió a l'examen, ordenada
de major a menor.

Compara aquest rànquing amb el que recordeu d'Iris. És el mateix?

*Com saps que ho has fet bé: el bosc aleatori arriba al 100 %; SVM i regressió
logística queden a prop del 98 %; l'arbre sol al 96,3 %; k-NN, que amb Iris anava
capdavanter, aquí tanca la taula amb un 94,4 %.*

```python
# Exercici 5
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

models = {
    "k-NN": KNeighborsClassifier(n_neighbors=5),
    "Arbre de decisió": DecisionTreeClassifier(random_state=42),
    "Bosc aleatori": RandomForestClassifier(n_estimators=100, random_state=42),
    "Regressió logística": LogisticRegression(random_state=42, max_iter=10000),
    "SVM": SVC(random_state=42),
}

resultats = []
for nom, model in models.items():
    model.fit(X_entrena_esc, y_entrena)
    precisio = model.score(X_examen_esc, y_examen)
    resultats.append((nom, precisio))

taula = pd.DataFrame(resultats, columns=["model", "precisio"]).sort_values(
    "precisio", ascending=False
)
taula

# A Iris el rànquing sol posar k-NN i SVM al capdavant; aquí el bosc aleatori
# guanya clarament i el k-NN, que a Iris anava bé, és qui menys precisió treu.
```

### Exercici 6 — Profunditat de l'arbre

Sobre les dades **sense escalar** (els arbres no en necessiten), entrena un
`DecisionTreeClassifier(max_depth=d, random_state=42)` per a cada `d` d'1 a 15.
Guarda la precisió a l'entrenament i a l'examen de cada arbre i dibuixa les dues
corbes juntes. A partir de quina profunditat comença el sobreajust (la precisió
d'entrenament segueix pujant però la d'examen ja no)?

*Com saps que ho has fet bé: amb `max_depth=1` la precisió d'examen és d'un 61 %.
A partir de `max_depth=4` l'arbre arriba al 100 % amb l'entrenament, però l'examen es
queda clavat al 96,3 % — la resta de profunditat és sobreajustament pur.*

```python
# Exercici 6
import matplotlib.pyplot as plt

profunditats = range(1, 16)
precisions_entrena = []
precisions_examen = []

for d in profunditats:
    model_d = DecisionTreeClassifier(max_depth=d, random_state=42)
    model_d.fit(X_entrena, y_entrena)
    precisions_entrena.append(model_d.score(X_entrena, y_entrena))
    precisions_examen.append(model_d.score(X_examen, y_examen))

plt.figure(figsize=(9, 5))
plt.plot(list(profunditats), precisions_entrena, marker="o", label="Entrenament")
plt.plot(list(profunditats), precisions_examen, marker="o", label="Examen")
plt.xlabel("Profunditat màxima (max_depth)")
plt.ylabel("Precisió")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print(f"max_depth=1:  entrenament {precisions_entrena[0]:.1%}, "
      f"examen {precisions_examen[0]:.1%}")
print(f"max_depth=4:  entrenament {precisions_entrena[3]:.1%}, "
      f"examen {precisions_examen[3]:.1%}")
print(f"max_depth=15: entrenament {precisions_entrena[-1]:.1%}, "
      f"examen {precisions_examen[-1]:.1%}")
```

### Exercici 7 — Importància de les variables

Entrena un `RandomForestClassifier(n_estimators=100, random_state=42)` amb les dades
sense escalar (els boscos tampoc en necessiten) i mira `feature_importances_`. Fes
un gràfic de barres horitzontal amb les 13 columnes ordenades de més a menys
importància. Quines tres manen?

*Com saps que ho has fet bé: `alcohol`, `color_intensity` i `flavanoids` són les tres
primeres, totes al voltant del 16 % d'importància cadascuna — i no coincideixen amb
les columnes que dominaven la distància de k-NN a l'exercici 4 (`proline`,
`magnesium`). Un bosc aleatori no treballa amb distàncies: l'escala no li afecta.*

```python
# Exercici 7
bosc_final = RandomForestClassifier(n_estimators=100, random_state=42)
bosc_final.fit(X_entrena, y_entrena)

importancies = pd.Series(
    bosc_final.feature_importances_, index=cols
).sort_values(ascending=False)

plt.figure(figsize=(8, 6))
importancies.plot(kind="barh")
plt.gca().invert_yaxis()
plt.xlabel("Importància")
plt.title("Importància de cada columna (bosc aleatori)")
plt.tight_layout()
plt.show()

print(importancies.head(3))
```
