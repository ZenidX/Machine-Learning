## Els objectes que manegues, i què els pots demanar

Les eines de la llista de dalt no treballen soltes: cadascuna et torna un
**objecte**, i el pas següent de l'exercici surt de demanar-li alguna cosa a aquell
objecte. Aquí tens, de cada un, el que faràs servir de veritat en aquest quadern:
els **atributs** (sense parèntesis) i els **mètodes** (amb parèntesis).

**El `Bunch` que torna `load_wine()`** — `sklearn.utils.Bunch`, una mena de
diccionari al qual també pots accedir amb punt (`wine.data` és el mateix que
`wine["data"]`):
- `.data` — les 178×13 mesures químiques
- `.target` — les 178 etiquetes, en números (0, 1, 2)
- `.feature_names` — la llista amb els 13 noms de columna; és el que hi ha a `cols`
- `.target_names` — els 3 noms de varietat; indexar-la amb `.target` és el que
  converteix els números en noms
- `.frame` — només amb `as_frame=True`: el `DataFrame` amb les 13 columnes **i**
  `target`
- `.DESCR` — el text que descriu el dataset (`print(wine.DESCR)` per llegir-lo)

**L'`ndarray` de NumPy** — el que hi ha dins de `X`, `X_entrena`, `X_examen`,
`X_entrena_esc`:
- `.shape` — la tupla (files, columnes). És el primer que has de mirar després de
  partir les dades
- `.dtype` — el tipus dels valors que hi ha a dins
- `.T` — la matriu transposada
- `.mean(axis=0)` i `.std(axis=0)` — mitjana i desviació **per columna**; amb
  `axis=1` te les fa per fila, i sense `axis` te les fa de tots els valors alhora
- `.argmax()` — la **posició** del valor més gran, no el valor
- `.reshape()` — canvia la forma sense tocar els valors

**El `DataFrame` i la `Series` de pandas** — `dades`, `desviacions`, `importancies`,
`taula`:
- del `DataFrame`: `.shape`, `.columns`, `.dtypes`, `.index` (atributs) i
  `.head()`, `.describe()`, `.drop(columns=[...])`, `.sort_values(...)`,
  `.to_numpy()` (mètodes)
- de la `Series`: `.value_counts()`, `.sort_values()`, `.plot(kind="barh")`
- `dades[cols]` et torna un `DataFrame` amb aquestes columnes; `dades["classe"]`
  et torna una sola columna com a `Series`

**Un model de scikit-learn** — `KNeighborsClassifier`, `DecisionTreeClassifier`,
`RandomForestClassifier`, `LogisticRegression`, `SVC`. Els cinc tenen **la mateixa
interfície**, i això és mitja assignatura:
- `.fit(X, y)` — entrena amb aquestes dades; torna el model mateix
- `.predict(X)` — un array amb una etiqueta predita per fila
- `.score(X, y)` — l'exactitud sobre aquestes dades, en un sol número
- `.classes_` — les classes que ha vist en entrenar, ordenades
- `.n_features_in_` — quantes columnes esperava; si li dones un nombre diferent a
  `.predict()`, peta
- `.feature_importances_` — **només** els models d'arbre (`DecisionTreeClassifier`,
  `RandomForestClassifier`): un pes per columna, i tots junts sumen 1
- `.coef_` — **només** `LogisticRegression` (i un `SVC` lineal): un coeficient per
  columna i classe
- `.get_depth()` i `.get_n_leaves()` — només l'arbre

**L'`StandardScaler`** — no és un model, és un **transformador**, i per això té uns
altres mètodes:
- `.fit(X)`, `.transform(X)`, i els dos de cop amb `.fit_transform(X)`. No té
  `.predict()` ni `.score()`
- `.mean_` i `.scale_` — la mitjana i la desviació de cada columna que ha après de
  `X_entrena`. Són exactament les que fa servir després a `.transform(X_examen)`, i
  és per això que `X_examen_esc` no et surt amb mitjana 0

Dues regles que valen per a tot el quadern, perquè són les que fan petar el codi
sense que sembli que hi hagi res malament:

- **Un atribut va sense parèntesis i un mètode amb parèntesis.** `model.coef_` et
  dona els coeficients; `model.coef_()` dona `TypeError`, perquè no és una funció.
  I a l'inrevés: el `.describe` d'un DataFrame sense parèntesis no calcula res,
  t'ensenya el mètode.
- **El guió baix al final vol dir «això ho he après de les dades».** `.coef_`,
  `.feature_importances_` i `.classes_` **no existeixen abans de cridar `.fit()`**:
  si els demanes abans, salta un error. El mateix passa amb els atributs apresos
  de qualsevol transformador.

Separa `dades` en `X` (les 13 columnes químiques) i `y` (la columna `classe`). Fes
el split 70/30 amb `train_test_split(..., test_size=0.3, random_state=42,

*Com saps que ho has fet bé: comprova la partició abans de creure't cap número.
`X_entrena` ha de tenir 124 files i `X_examen` 54, amb 13 columnes totes dues, i
124 + 54 ha de fer 178. Comprova també que `stratify` ha fet la seva feina: les
proporcions de classe de `y_examen` (`y_examen.value_counts(normalize=True)`) han de
quedar molt a prop de les de `y` sencer. Si la partició està torta, cap exactitud que
mesuris després no es pot comparar amb res. L'exactitud que t'hagi sortit a l'examen,
apunta-la: la faràs servir com a referència a la resta del quadern.*

Abans de començar, la idea en un dibuix: què fa `stratify`. Cent etiquetes inventades de
tres classes, partides 70/30 dues vegades. Sense `stratify` l'examen pot quedar amb una
proporció de classes diferent de la del conjunt sencer; amb `stratify` les tres barres
queden iguals. És la comprovació que us demanem amb `value_counts(normalize=True)`.

```python
from sklearn.model_selection import train_test_split

y_il = np.array(["A"] * 50 + ["B"] * 35 + ["C"] * 15)   # etiquetes inventades
X_il = np.arange(100).reshape(-1, 1)
_, _, _, ye_sense = train_test_split(X_il, y_il, test_size=0.3, random_state=3)
_, _, _, ye_amb = train_test_split(X_il, y_il, test_size=0.3, random_state=3, stratify=y_il)

def proporcions_il(v):
    return [np.mean(v == c) for c in ["A", "B", "C"]]

fig, ax = plt.subplots(figsize=(9, 4))
amplada = 0.27
for i, (nom, v, color) in enumerate([("conjunt sencer", y_il, "tab:blue"),
                                     ("examen sense stratify", ye_sense, "tab:red"),
                                     ("examen amb stratify", ye_amb, "tab:green")]):
    ax.bar(np.arange(3) + (i - 1) * amplada, proporcions_il(v), amplada, color=color, label=nom)
ax.set_xticks(range(3), ["classe A", "classe B", "classe C"])
ax.set_ylabel("proporció")
ax.set_title("Dades inventades: proporcions de classe a l'examen")
ax.legend()
ax.grid(alpha=0.3, axis="y")
plt.show()
```

Compara les dues exactituds i escriu en un comentari, amb les teves paraules, què

*Com saps que ho has fet bé: comprova que l'escalat s'ha aplicat de veritat.
`X_entrena_esc.mean(axis=0)` ha de donar pràcticament zero a cada columna, i
`X_entrena_esc.std(axis=0)` pràcticament u. A `X_examen_esc` no hi sortiran

l'entrenament les mitjanes surten 0 i les desviacions 1 exactes; a l'examen, a prop però

### Exercici 4 — Quina columna té la culpa

Calcula la desviació estàndard de cada columna de `X_entrena` (sense escalar) i
ordena-les de més gran a més petita. Aquestes són, en unitats originals, les
columnes que dominaran (o no) el càlcul de distàncies de k-NN abans d'escalar.

*Com saps que ho has fet bé: la `Series` ha de tenir 13 valors, un per columna,
cap de negatiu, i el primer ha de ser el més gran (`desviacions.iloc[0] ==
desviacions.max()`). Comprova'n un a mà: `X_entrena[:, i].std()` per a una sola columna
ha de donar el mateix número que la `Series` per a aquella columna. I un ordre de
magnitud: la desviació d'una columna no pot passar de la meitat del seu rang (màxim −
mínim) a `X_entrena`; si et surt més gran, l'has calculada per files (`axis=1`) en lloc
de per columnes. Quines columnes encapçalen la llista és la teva resposta.*

Abans de començar, la idea en un dibuix: per què la desviació estàndard d'una columna diu
quant pesarà a la distància. Dues columnes inventades: per a moltes parelles de files, la
diferència $|a_i - b_i|$ en cadascuna. La columna amb més desviació dona diferències més
grans, i a la distància aquestes diferències s'eleven al quadrat.

```python
rng = np.random.default_rng(6)
gran_il = rng.normal(500, 150, 300)                      # desviació ~150
petita_il = rng.normal(1.5, 0.4, 300)                    # desviació ~0.4
i_il, j_il = rng.integers(0, 300, 1000), rng.integers(0, 300, 1000)

dif_gran_il = np.abs(gran_il[i_il] - gran_il[j_il])
dif_petita_il = np.abs(petita_il[i_il] - petita_il[j_il])
calaixos_il = np.linspace(0, 700, 71)                  # els mateixos calaixos per a totes dues

fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(dif_gran_il, bins=calaixos_il, color="tab:orange", alpha=0.7,
        label=f"columna amb desviació {gran_il.std():.0f}")
ax.hist(dif_petita_il, bins=calaixos_il, color="tab:blue", alpha=0.9,
        label=f"columna amb desviació {petita_il.std():.1f}")
ax.set_ylim(0, 140)
ax.annotate(f"les {len(dif_petita_il)} diferències de la columna petita\ncauen totes aquí (cap passa de {dif_petita_il.max():.1f})",
            (5, 135), xytext=(150, 110), arrowprops=dict(arrowstyle="->", color="tab:blue"), color="tab:blue")
ax.set_xlabel("diferència entre dues files, |a - b|")
ax.set_ylabel("parelles")
ax.set_title("Dades inventades: la desviació mana la mida de les diferències")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

### Exercici 6 — Profunditat de l'arbre

Sobre les dades **sense escalar** (els arbres no en necessiten), entrena un
`DecisionTreeClassifier(max_depth=d, random_state=42)` per a cada `d` d'1 a 15.
Guarda l'exactitud a l'entrenament i a l'examen de cada arbre i dibuixa les dues
corbes juntes. A partir de quina profunditat comença el sobreajust (l'exactitud
d'entrenament segueix pujant però la d'examen ja no)?

*Com saps que ho has fet bé: dues comprovacions d'estructura, cap número d'exactitud.
Primera: la corba d'entrenament no pot baixar mai quan puja la profunditat —més
profunditat és més llibertat per encertar el que ja ha vist—, i a partir d'una certa
`d` les dues corbes es queden planes, perquè l'arbre ja ha crescut tot el que podia i
afegir-hi `max_depth` no canvia res. Si te'n baixa cap tram, tens un error al bucle.
Segona: amb `max_depth=1` l'arbre té dues fulles per a tres classes, així que hi ha
una classe que no pot predir ni una vegada; comprova-ho amb `confusion_matrix` (de
`sklearn.metrics`) i hi trobaràs una columna sencera a zero. On comença el
sobreajust, i quant de marge et deixa, ho llegeixes tu al gràfic.*

Abans de començar, la idea en un dibuix: què vol dir `max_depth`. Punts inventats de tres
classes i el color que posa l'arbre a cada zona del pla, amb profunditat 1, 3 i sense
límit. Amb `max_depth=1` només hi ha un tall, dues fulles i, per tant, una classe que no
surt mai. Sense límit, l'arbre talla fins a encerclar cada punt solt. La corba que
dibuixareu mesura on passa de l'un a l'altre amb Wine.

```python
from sklearn.datasets import make_blobs
from sklearn.tree import DecisionTreeClassifier
from matplotlib.colors import ListedColormap

Xb_il, yb_il = make_blobs(n_samples=150, centers=[[0, 0], [4, 0], [2, 3.5]],
                          cluster_std=1.5, random_state=0)  # inventades
gx_il, gy_il = np.meshgrid(np.linspace(Xb_il[:, 0].min() - 1, Xb_il[:, 0].max() + 1, 300),
                           np.linspace(Xb_il[:, 1].min() - 1, Xb_il[:, 1].max() + 1, 300))
graella_il = np.column_stack([gx_il.ravel(), gy_il.ravel()])
fons_il = ListedColormap(["#c6dbef", "#fdd0a2", "#c7e9c0"])
punts_col = ["tab:blue", "tab:orange", "tab:green"]

fig, axs = plt.subplots(1, 3, figsize=(15, 4.6), sharex=True, sharey=True)
for ax, prof in zip(axs, [1, 3, None]):
    arbre_il = DecisionTreeClassifier(max_depth=prof, random_state=0).fit(Xb_il, yb_il)
    ax.contourf(gx_il, gy_il, arbre_il.predict(graella_il).reshape(gx_il.shape),
                levels=[-0.5, 0.5, 1.5, 2.5], cmap=fons_il)
    for c in range(3):
        ax.scatter(Xb_il[yb_il == c, 0], Xb_il[yb_il == c, 1], color=punts_col[c], edgecolor="k", s=18)
    ax.set_title(f"max_depth={prof}\n{arbre_il.get_n_leaves()} fulles · "
                 f"{len(np.unique(arbre_il.predict(Xb_il)))} classes predites")
plt.tight_layout()
plt.show()
```

Entrena un `RandomForestClassifier(n_estimators=100, random_state=42)` amb les dades

*Com saps que ho has fet bé: les 13 importàncies han de sumar 1
(`importancies.sum()`), cap no pot ser negativa, i al gràfic la barra més llarga ha de
quedar a dalt de tot (si queda a baix, has ordenat al revés). Com que un bosc aleatori no
treballa amb distàncies, l'escala no li afecta: si el reentrenes amb `X_entrena_esc`, les

Abans de començar, la idea en un dibuix: com es llegeix `feature_importances_`. Cinc
importàncies inventades: sumen 1, i en un `barh` la primera fila de la `Series` queda
**a baix**. Per això, si ordeneu de petita a gran amb `.sort_values()`, la més important
surt a dalt de tot.

```python
import pandas as pd

imp_il = pd.Series([0.05, 0.40, 0.10, 0.30, 0.15],
                   index=["col_a", "col_b", "col_c", "col_d", "col_e"])   # inventades

fig, axs = plt.subplots(1, 2, figsize=(12, 3.8))
imp_il.plot(kind="barh", ax=axs[0], color="lightgray", edgecolor="gray")
axs[0].set_title("Sense ordenar")
imp_il.sort_values().plot(kind="barh", ax=axs[1], color="tab:blue")
axs[1].set_title(f"Amb .sort_values() (suma = {imp_il.sum():.2f})")
for ax in axs:
    ax.set_xlabel("importància (inventada)")
    ax.grid(alpha=0.3, axis="x")
plt.tight_layout()
plt.show()
```
